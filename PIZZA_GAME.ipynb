```python
# ============================================================
# 🍕 FRACTION PIZZA GAME
# LEVEL 3, LEVEL 4 & LEVEL 5
# Google Colab Prototype
# ============================================================

import random
import math
import io

from fractions import Fraction
from PIL import Image, ImageDraw
from IPython.display import display, clear_output
import ipywidgets as widgets


# ============================================================
# GLOBAL GAME VARIABLES
# ============================================================

score = 0
lives = 3
streak = 0

current_level = 3

# Used for selected pizza pieces
selected_pieces = []

# ============================================================
# COLORS
# ============================================================

PIZZA_COLOR = "#FFD966"
SELECTED_COLOR = "#90EE90"
CRUST_COLOR = "#C97A40"
PEPPERONI_COLOR = "#D94A4A"
LINE_COLOR = "#8B5A2B"


# ============================================================
# CREATE PIZZA IMAGE
# ============================================================

def create_pizza(total_pieces, selected=None, size=350):

    if selected is None:
        selected = []

    selected = set(selected)

    img = Image.new("RGB", (size, size), "white")
    draw = ImageDraw.Draw(img)

    margin = 25
    box = (
        margin,
        margin,
        size - margin,
        size - margin
    )

    # --------------------------------------------------------
    # Pizza base
    # --------------------------------------------------------

    draw.ellipse(
        box,
        fill=CRUST_COLOR,
        outline="#6B3E26",
        width=5
    )

    inner_margin = 12

    inner_box = (
        margin + inner_margin,
        margin + inner_margin,
        size - margin - inner_margin,
        size - margin - inner_margin
    )

    # --------------------------------------------------------
    # Draw individual slices
    # --------------------------------------------------------

    slice_angle = 360 / total_pieces

    for i in range(total_pieces):

        start_angle = i * slice_angle
        end_angle = (i + 1) * slice_angle

        if i in selected:
            fill_color = SELECTED_COLOR
        else:
            fill_color = PIZZA_COLOR

        draw.pieslice(
            inner_box,
            start=start_angle,
            end=end_angle,
            fill=fill_color,
            outline=LINE_COLOR
        )

    # --------------------------------------------------------
    # Add pepperoni
    # --------------------------------------------------------

    center_x = size // 2
    center_y = size // 2

    radius = (size - 2 * margin) // 2

    for i in range(total_pieces):

        angle = math.radians(
            (i + 0.5) * slice_angle
        )

        x = center_x + int(radius * 0.48 * math.cos(angle))
        y = center_y + int(radius * 0.48 * math.sin(angle))

        r = 9

        draw.ellipse(
            (
                x-r,
                y-r,
                x+r,
                y+r
            ),
            fill=PEPPERONI_COLOR
        )

    # --------------------------------------------------------
    # Center point
    # --------------------------------------------------------

    draw.ellipse(
        (
            center_x - 5,
            center_y - 5,
            center_x + 5,
            center_y + 5
        ),
        fill="#8B4513"
    )

    # --------------------------------------------------------
    # Convert image to PNG bytes
    # --------------------------------------------------------

    buffer = io.BytesIO()
    img.save(buffer, format="PNG")

    return buffer.getvalue()


# ============================================================
# WIDGETS
# ============================================================

pizza_output = widgets.Output()

status_label = widgets.HTML()

question_label = widgets.HTML()

feedback_label = widgets.HTML()

score_label = widgets.HTML()

answer_box = widgets.VBox()

next_button = widgets.Button(
    description="➡️ Next Level",
    button_style="success",
    layout=widgets.Layout(
        width="180px",
        height="40px"
    )
)

next_button.layout.display = "none"


# ============================================================
# UPDATE STATUS
# ============================================================

def update_status():

    score_label.value = f"""
    <div style="
        font-size:18px;
        font-weight:bold;
        padding:10px;
        background:#F2F2F2;
        border-radius:10px;
        text-align:center;
    ">
        ⭐ Score: {score}
        &nbsp;&nbsp;&nbsp;
        ❤️ Lives: {lives}
        &nbsp;&nbsp;&nbsp;
        🔥 Streak: {streak}
    </div>
    """


# ============================================================
# DISPLAY PIZZA
# ============================================================

def show_pizza(total_pieces, selected):

    pizza_output.clear_output(wait=True)

    image_data = create_pizza(
        total_pieces,
        selected
    )

    with pizza_output:

        display(
            widgets.Image(
                value=image_data,
                format="png",
                width=350,
                height=350
            )
        )


# ============================================================
# RESET FEEDBACK
# ============================================================

def reset_feedback():

    feedback_label.value = ""

    next_button.layout.display = "none"


# ============================================================
# GAME OVER CHECK
# ============================================================

def check_game_over():

    if lives <= 0:

        feedback_label.value = """
        <div style="
            background:#FFCCCC;
            padding:15px;
            border-radius:10px;
            text-align:center;
            font-size:22px;
            font-weight:bold;
        ">
        💔 GAME OVER! <br><br>
        You used all 3 lives.
        </div>
        """

        answer_box.children = []

        return True

    return False


# ============================================================
# LEVEL 3
# FRACTION DETECTIVE
# ============================================================

def level3():

    global selected_pieces
    global score
    global lives
    global streak

    reset_feedback()

    selected_pieces = []

    total = random.choice([4, 8])

    selected_count = random.randint(
        1,
        total - 1
    )

    selected_pieces = random.sample(
        range(total),
        selected_count
    )

    correct_fraction = Fraction(
        selected_count,
        total
    )

    # --------------------------------------------------------
    # Header
    # --------------------------------------------------------

    question_label.value = f"""
    <h2>🔢 Level 3 – Fraction Detective</h2>

    <p style="font-size:18px;">
    Look carefully at the pizza.
    </p>

    <p style="font-size:20px;font-weight:bold;">
    🍕 What fraction of the pizza is selected?
    </p>
    """

    status_label.value = f"""
    <div style="
        text-align:center;
        font-size:18px;
        color:#555;
    ">
    🕵️ Count the highlighted pieces!
    </div>
    """

    show_pizza(
        total,
        selected_pieces
    )

    update_status()

    # --------------------------------------------------------
    # Generate answer choices
    # --------------------------------------------------------

    choices = set()

    choices.add(
        f"{selected_count}/{total}"
    )

    # Generate incorrect choices
    while len(choices) < 4:

        n = random.randint(1, total)

        choice = f"{n}/{total}"

        choices.add(choice)

    choices = list(choices)
    random.shuffle(choices)

    buttons = []

    for choice in choices:

        button = widgets.Button(
            description=choice,
            layout=widgets.Layout(
                width="120px",
                height="45px"
            )
        )

        def check_answer(
            b,
            choice=choice
        ):

            global score
            global lives
            global streak

            selected_fraction = Fraction(
                int(choice.split("/")[0]),
                int(choice.split("/")[1])
            )

            if selected_fraction == correct_fraction:

                score += 20
                streak += 1

                # 3 correct answers bonus
                if streak == 3:

                    score += 20

                    feedback_label.value = """
                    <div style="
                        background:#D9FFD9;
                        padding:15px;
                        border-radius:10px;
                        font-size:20px;
                        text-align:center;
                    ">
                    🎉 Correct! +20 points<br>
                    🔥 3-answer streak bonus +20!
                    </div>
                    """

                    streak = 0

                else:

                    feedback_label.value = f"""
                    <div style="
                        background:#D9FFD9;
                        padding:15px;
                        border-radius:10px;
                        font-size:20px;
                        text-align:center;
                    ">
                    ✅ Correct!<br>
                    The answer is <b>{selected_count}/{total}</b><br>
                    ⭐ +20 points
                    </div>
                    """

                update_status()

                for btn in buttons:
                    btn.disabled = True

                next_button.layout.display = "block"

            else:

                lives -= 1
                streak = 0

                feedback_label.value = f"""
                <div style="
                    background:#FFD6D6;
                    padding:15px;
                    border-radius:10px;
                    font-size:20px;
                    text-align:center;
                ">
                ❌ Try Again!<br>
                You selected <b>{choice}</b>.
                </div>
                """

                update_status()

                if check_game_over():

                    for btn in buttons:
                        btn.disabled = True

        button.on_click(check_answer)

        buttons.append(button)

    answer_box.children = [
        widgets.HBox(
            buttons[:2],
            layout=widgets.Layout(
                justify_content="center"
            )
        ),
        widgets.HBox(
            buttons[2:],
            layout=widgets.Layout(
                justify_content="center"
            )
        )
    ]


# ============================================================
# LEVEL 4
# EQUIVALENT FRACTIONS
# ============================================================

def level4():

    global score
    global lives
    global streak

    reset_feedback()

    question_label.value = """
    <h2>🔄 Level 4 – Equivalent Fractions</h2>

    <p style="font-size:20px;">
    Look at the shaded part of the pizza.
    </p>

    <p style="font-size:22px;font-weight:bold;">
    🍕 Which fraction is equal to <b>1/2</b>?
    </p>
    """

    status_label.value = """
    <div style="
        text-align:center;
        font-size:18px;
    ">
    💡 Equivalent fractions represent the same amount.
    </div>
    """

    # --------------------------------------------------------
    # Show 1/2 pizza
    # --------------------------------------------------------

    show_pizza(
        2,
        [0]
    )

    update_status()

    choices = [
        "1/4",
        "2/4",
        "3/8",
        "1/8"
    ]

    random.shuffle(choices)

    buttons = []

    for choice in choices:

        button = widgets.Button(
            description=choice,
            layout=widgets.Layout(
                width="120px",
                height="45px"
            )
        )

        def check_answer(
            b,
            choice=choice
        ):

            global score
            global lives
            global streak

            if choice == "2/4":

                score += 20
                streak += 1

                feedback_label.value = """
                <div style="
                    background:#D9FFD9;
                    padding:15px;
                    border-radius:10px;
                    font-size:20px;
                    text-align:center;
                ">
                🎉 Correct!<br><br>

                <b>1/2 = 2/4 = 4/8</b>

                <br><br>
                They all represent the same part of a pizza.

                <br><br>
                ⭐ +20 points
                </div>
                """

                update_status()

                for btn in buttons:
                    btn.disabled = True

                next_button.layout.display = "block"

            else:

                lives -= 1
                streak = 0

                feedback_label.value = """
                <div style="
                    background:#FFD6D6;
                    padding:15px;
                    border-radius:10px;
                    font-size:20px;
                    text-align:center;
                ">
                ❌ Not quite!
                <br><br>
                Look at how much of the pizza is shaded.
                <br>
                Try to find the fraction representing the same amount.
                </div>
                """

                update_status()

                check_game_over()

        button.on_click(check_answer)

        buttons.append(button)

    answer_box.children = [
        widgets.HBox(
            buttons[:2],
            layout=widgets.Layout(
                justify_content="center"
            )
        ),
        widgets.HBox(
            buttons[2:],
            layout=widgets.Layout(
                justify_content="center"
            )
        )
    ]


# ============================================================
# LEVEL 5
# COMPARE THE PIZZAS
# ============================================================

def level5():

    global score
    global lives
    global streak

    reset_feedback()

    # --------------------------------------------------------
    # Random comparison
    # --------------------------------------------------------

    comparisons = [

        (2, 4, 1, 2),
        (1, 2, 1, 4),
        (3, 8, 1, 4),
        (2, 8, 1, 4),
        (3, 4, 1, 2),
        (1, 4, 2, 4)
    ]

    a_num, a_den, b_num, b_den = random.choice(
        comparisons
    )

    fraction_a = Fraction(
        a_num,
        a_den
    )

    fraction_b = Fraction(
        b_num,
        b_den
    )

    if fraction_a > fraction_b:
        correct = ">"

    elif fraction_a < fraction_b:
        correct = "<"

    else:
        correct = "="

    # --------------------------------------------------------
    # Question
    # --------------------------------------------------------

    question_label.value = f"""
    <h2>⚖️ Level 5 – Compare the Pizzas</h2>

    <p style="font-size:20px;">
    Which pizza has more?
    </p>

    <p style="font-size:24px;font-weight:bold;">
    🍕 {a_num}/{a_den}
    &nbsp;&nbsp;&nbsp; ? &nbsp;&nbsp;&nbsp;
    {b_num}/{b_den} 🍕
    </p>
    """

    status_label.value = """
    <div style="
        text-align:center;
        font-size:18px;
    ">
    👀 Compare the shaded parts of the two pizzas.
    </div>
    """

    # --------------------------------------------------------
    # Create two pizza images
    # --------------------------------------------------------

    img_a = widgets.Image(
        value=create_pizza(
            a_den,
            list(range(a_num))
        ),
        format="png",
        width=300,
        height=300
    )

    img_b = widgets.Image(
        value=create_pizza(
            b_den,
            list(range(b_num))
        ),
        format="png",
        width=300,
        height=300
    )

    pizza_output.clear_output(wait=True)

    with pizza_output:

        display(
            widgets.HBox(
                [
                    widgets.VBox(
                        [
                            widgets.HTML(
                                f"<h3>Pizza A<br>{a_num}/{a_den}</h3>"
                            ),
                            img_a
                        ],
                        layout=widgets.Layout(
                            align_items="center"
                        )
                    ),

                    widgets.HTML(
                        "<h1 style='padding:120px 20px;'>?</h1>"
                    ),

                    widgets.VBox(
                        [
                            widgets.HTML(
                                f"<h3>Pizza B<br>{b_num}/{b_den}</h3>"
                            ),
                            img_b
                        ],
                        layout=widgets.Layout(
                            align_items="center"
                        )
                    )
                ],
                layout=widgets.Layout(
                    justify_content="center"
                )
            )
        )

    update_status()

    # --------------------------------------------------------
    # Comparison buttons
    # --------------------------------------------------------

    choices = [
        ">",
        "<",
        "="
    ]

    buttons = []

    for choice in choices:

        button = widgets.Button(
            description=choice,
            layout=widgets.Layout(
                width="90px",
                height="55px"
            )
        )

        def check_answer(
            b,
            choice=choice
        ):

            global score
            global lives
            global streak

            if choice == correct:

                score += 30
                streak += 1

                if correct == "=":

                    explanation = f"""
                    <b>{a_num}/{a_den} = {b_num}/{b_den}</b>
                    <br><br>
                    Both pizzas have the same shaded area.
                    """

                elif correct == ">":

                    explanation = f"""
                    <b>{a_num}/{a_den} &gt; {b_num}/{b_den}</b>
                    <br><br>
                    Pizza A has more shaded area.
                    """

                else:

                    explanation = f"""
                    <b>{a_num}/{a_den} &lt; {b_num}/{b_den}</b>
                    <br><br>
                    Pizza B has more shaded area.
                    """

                feedback_label.value = f"""
                <div style="
                    background:#D9FFD9;
                    padding:15px;
                    border-radius:10px;
                    font-size:20px;
                    text-align:center;
                ">
                🎉 Correct!
                <br><br>

                {explanation}

                <br>
                ⭐ +30 points
                </div>
                """

                update_status()

                for btn in buttons:
                    btn.disabled = True

                next_button.layout.display = "block"

            else:

                lives -= 1
                streak = 0

                feedback_label.value = """
                <div style="
                    background:#FFD6D6;
                    padding:15px;
                    border-radius:10px;
                    font-size:20px;
                    text-align:center;
                ">
                ❌ Try again!
                <br><br>
                Compare the shaded areas carefully.
                </div>
                """

                update_status()

                check_game_over()

        button.on_click(check_answer)

        buttons.append(button)

    answer_box.children = [
        widgets.HBox(
            buttons,
            layout=widgets.Layout(
                justify_content="center"
            )
        )
    ]


# ============================================================
# NEXT LEVEL BUTTON
# ============================================================

def next_level_clicked(b):

    global current_level

    next_button.layout.display = "none"

    current_level += 1

    if current_level == 4:

        level4()

    elif current_level == 5:

        level5()

    else:

        feedback_label.value = """
        <div style="
            background:#E6F2FF;
            padding:20px;
            border-radius:10px;
            text-align:center;
            font-size:22px;
        ">
        🎉 Levels 3–5 completed!
        <br><br>
        Ready to continue with the next Fraction Pizza level.
        </div>
        """

        answer_box.children = []


next_button.on_click(next_level_clicked)


# ============================================================
# MAIN GAME DISPLAY
# ============================================================

game_ui = widgets.VBox(
    [
        widgets.HTML(
            """
            <div style="
                text-align:center;
                background:#FFCC66;
                padding:15px;
                border-radius:15px;
            ">
            <h1>🍕 FRACTION PIZZA</h1>
            <p style="font-size:18px;">
            Learn Fractions • Build • Play • Challenge
            </p>
            </div>
            """
        ),

        score_label,

        question_label,

        status_label,

        pizza_output,

        answer_box,

        feedback_label,

        next_button
    ],
    layout=widgets.Layout(
        width="100%",
        align_items="center"
    )
)


# ============================================================
# START LEVEL 3
# ============================================================

display(game_ui)

level3()
```

```python
# ============================================================
# ➕ LEVEL 6 – PIZZA FRACTION ADDITION
# ============================================================

def level6():

    global score
    global lives
    global streak
    global selected_pieces

    reset_feedback()

    # --------------------------------------------------------
    # Fixed example for this level
    # --------------------------------------------------------

    total_pieces = 4

    first_fraction = Fraction(1, 4)
    second_fraction = Fraction(1, 4)

    required_pieces = 2

    selected_pieces = []

    # --------------------------------------------------------
    # Question
    # --------------------------------------------------------

    question_label.value = """
    <h2>➕ Level 6 – Pizza Fraction Addition</h2>

    <p style="font-size:20px;">
    You ate <b>1/4</b> of a pizza in the afternoon
    and another <b>1/4</b> in the evening.
    </p>

    <p style="font-size:22px;font-weight:bold;">
    🍕 How much pizza did you eat altogether?
    </p>
    """

    status_label.value = """
    <div style="
        text-align:center;
        font-size:18px;
    ">
    👆 Select <b>2 pieces</b> from the pizza.
    <br>
    1 piece + 1 piece = 2 pieces
    </div>
    """

    update_status()

    # --------------------------------------------------------
    # Pizza image
    # --------------------------------------------------------

    show_pizza(
        total_pieces,
        selected_pieces
    )

    # --------------------------------------------------------
    # Create pizza-piece buttons
    # --------------------------------------------------------

    piece_buttons = []

    for i in range(total_pieces):

        button = widgets.Button(
            description=f"🍕 {i+1}",
            layout=widgets.Layout(
                width="100px",
                height="45px"
            )
        )

        def select_piece(
            b,
            index=i
        ):

            global selected_pieces

            # ------------------------------------------------
            # Select / deselect piece
            # ------------------------------------------------

            if index in selected_pieces:

                selected_pieces.remove(index)

                b.button_style = ""

            else:

                if len(selected_pieces) < required_pieces:

                    selected_pieces.append(index)

                    b.button_style = "success"

            # ------------------------------------------------
            # Update pizza
            # ------------------------------------------------

            show_pizza(
                total_pieces,
                selected_pieces
            )

            status_label.value = f"""
            <div style="
                text-align:center;
                font-size:18px;
            ">
            🟩 Selected pieces:
            <b>{len(selected_pieces)}</b> / {required_pieces}
            </div>
            """

            # ------------------------------------------------
            # Enable check button when 2 pieces selected
            # ------------------------------------------------

            if len(selected_pieces) == required_pieces:

                check_button.disabled = False

            else:

                check_button.disabled = True

        button.on_click(select_piece)

        piece_buttons.append(button)

    # --------------------------------------------------------
    # Check button
    # --------------------------------------------------------

    check_button = widgets.Button(
        description="✅ Add Fractions",
        button_style="primary",
        disabled=True,
        layout=widgets.Layout(
            width="180px",
            height="45px"
        )
    )

    def check_addition(b):

        global score
        global streak

        # ----------------------------------------------------
        # Correct selection
        # ----------------------------------------------------

        if len(selected_pieces) == required_pieces:

            score += 20
            streak += 1

            feedback_label.value = """
            <div style="
                background:#D9FFD9;
                padding:20px;
                border-radius:12px;
                text-align:center;
                font-size:21px;
            ">

            🎉 <b>Excellent!</b>

            <br><br>

            You selected 2 pieces out of 4.

            <br><br>

            🍕 <b>1/4 + 1/4 = 2/4</b>

            <br><br>

            And:

            <br>

            <b>2/4 = 1/2</b>

            <br><br>

            ⭐ <b>+20 points</b>

            <br><br>

            You ate <b>half of the pizza!</b> 🍕

            </div>
            """

            update_status()

            # Disable all buttons
            for btn in piece_buttons:
                btn.disabled = True

            check_button.disabled = True

            next_button.layout.display = "block"

        else:

            feedback_label.value = """
            <div style="
                background:#FFF2CC;
                padding:15px;
                border-radius:10px;
                text-align:center;
                font-size:20px;
            ">
            👆 Select two pizza pieces.
            </div>
            """

    check_button.on_click(check_addition)

    # --------------------------------------------------------
    # Display piece buttons
    # --------------------------------------------------------

    answer_box.children = [

        widgets.HTML(
            "<h3>🍕 Select the pieces you ate:</h3>"
        ),

        widgets.HBox(
            piece_buttons,
            layout=widgets.Layout(
                justify_content="center"
            )
        ),

        check_button
    ]


# ============================================================
# UPDATE NEXT LEVEL BUTTON
# ============================================================

def next_level_clicked(b):

    global current_level

    next_button.layout.display = "none"

    current_level += 1

    if current_level == 4:

        level4()

    elif current_level == 5:

        level5()

    elif current_level == 6:

        level6()

    else:

        feedback_label.value = """
        <div style="
            background:#D9FFD9;
            padding:20px;
            border-radius:12px;
            text-align:center;
            font-size:22px;
        ">
        🎉 Level 6 completed!
        <br><br>
        You learned how to add pizza fractions.
        <br><br>
        <b>1/4 + 1/4 = 2/4 = 1/2</b>
        </div>
        """

        answer_box.children = []
```

```python
# ============================================================
# 🍕 LEVEL 6 – PIZZA FRACTION ADDITION
# Standalone Google Colab Version
# ============================================================

import math
import io

from PIL import Image, ImageDraw
from IPython.display import display
import ipywidgets as widgets


# ============================================================
# GAME VARIABLES
# ============================================================

level6_score = 0
level6_lives = 3
level6_selected = []


# ============================================================
# CREATE PIZZA
# ============================================================

def make_pizza(total=4, selected=None):

    if selected is None:
        selected = []

    selected = set(selected)

    size = 400

    img = Image.new(
        "RGB",
        (size, size),
        "white"
    )

    draw = ImageDraw.Draw(img)

    # Pizza circle
    margin = 30

    box = (
        margin,
        margin,
        size - margin,
        size - margin
    )

    # Crust
    draw.ellipse(
        box,
        fill="#C47A44",
        outline="#6B3E26",
        width=6
    )

    # Cheese area
    inner = (
        margin + 12,
        margin + 12,
        size - margin - 12,
        size - margin - 12
    )

    angle = 360 / total

    # --------------------------------------------------------
    # Draw slices
    # --------------------------------------------------------

    for i in range(total):

        start = i * angle
        end = (i + 1) * angle

        if i in selected:
            color = "#7CFC90"       # selected
        else:
            color = "#FFD966"       # normal

        draw.pieslice(
            inner,
            start=start,
            end=end,
            fill=color,
            outline="#8B5A2B",
            width=3
        )

    # --------------------------------------------------------
    # Pepperoni
    # --------------------------------------------------------

    center = size / 2
    radius = 130

    for i in range(total):

        theta = math.radians(
            (i + 0.5) * angle
        )

        x = center + radius * math.cos(theta)
        y = center + radius * math.sin(theta)

        r = 10

        draw.ellipse(
            (
                int(x-r),
                int(y-r),
                int(x+r),
                int(y+r)
            ),
            fill="#D94A4A"
        )

    # --------------------------------------------------------
    # Convert to PNG
    # --------------------------------------------------------

    buffer = io.BytesIO()

    img.save(
        buffer,
        format="PNG"
    )

    return buffer.getvalue()


# ============================================================
# IMAGE WIDGET
# ============================================================

pizza_image = widgets.Image(
    value=make_pizza(4, []),
    format="png",
    width=400,
    height=400
)


# ============================================================
# TEXT / OUTPUT WIDGETS
# ============================================================

title = widgets.HTML(
    """
    <div style="
        background:#FFCC66;
        padding:15px;
        border-radius:15px;
        text-align:center;
    ">

    <h1>🍕 Level 6 – Pizza Fraction Addition</h1>

    <p style="font-size:20px;">
    Learn to add fractions using pizza!
    </p>

    </div>
    """
)


question = widgets.HTML(
    """
    <div style="
        text-align:center;
        font-size:20px;
        padding:10px;
    ">

    🍕 You ate <b>1/4</b> of a pizza
    in the afternoon.

    <br><br>

    Then you ate another <b>1/4</b>
    in the evening.

    <br><br>

    <b>How much pizza did you eat altogether?</b>

    </div>
    """
)


status = widgets.HTML(
    """
    <div style="
        text-align:center;
        font-size:18px;
        padding:10px;
    ">
    👆 Select <b>2 pieces</b>.
    </div>
    """
)


feedback = widgets.HTML("")


score_display = widgets.HTML(
    """
    <div style="
        text-align:center;
        font-size:18px;
        font-weight:bold;
    ">
    ⭐ Score: 0 &nbsp;&nbsp;&nbsp; ❤️ Lives: 3
    </div>
    """
)


# ============================================================
# PIECE BUTTONS
# ============================================================

piece_buttons = []


# ============================================================
# UPDATE PIZZA
# ============================================================

def update_pizza():

    pizza_image.value = make_pizza(
        4,
        level6_selected
    )

    status.value = f"""
    <div style="
        text-align:center;
        font-size:18px;
        padding:10px;
    ">

    🟩 Selected:
    <b>{len(level6_selected)}/2</b>

    </div>
    """


# ============================================================
# UPDATE SCORE
# ============================================================

def update_score():

    score_display.value = f"""
    <div style="
        text-align:center;
        font-size:18px;
        font-weight:bold;
        padding:10px;
    ">

    ⭐ Score: {level6_score}
    &nbsp;&nbsp;&nbsp;
    ❤️ Lives: {level6_lives}

    </div>
    """


# ============================================================
# SELECT PIZZA PIECE
# ============================================================

def select_piece(button, index):

    global level6_selected

    # --------------------------------------------------------
    # Deselect
    # --------------------------------------------------------

    if index in level6_selected:

        level6_selected.remove(index)

        button.button_style = ""

    # --------------------------------------------------------
    # Select
    # --------------------------------------------------------

    else:

        if len(level6_selected) < 2:

            level6_selected.append(index)

            button.button_style = "success"

    update_pizza()

    # --------------------------------------------------------
    # Enable check button
    # --------------------------------------------------------

    if len(level6_selected) == 2:

        check_button.disabled = False

    else:

        check_button.disabled = True


# ============================================================
# CREATE 4 BUTTONS
# ============================================================

for i in range(4):

    btn = widgets.Button(
        description=f"🍕 Piece {i+1}",
        layout=widgets.Layout(
            width="130px",
            height="45px"
        )
    )

    # Important:
    # Use a lambda with default argument to remember index

    btn.on_click(
        lambda b, index=i:
        select_piece(b, index)
    )

    piece_buttons.append(btn)


# ============================================================
# CHECK BUTTON
# ============================================================

check_button = widgets.Button(
    description="✅ Check Addition",
    button_style="primary",
    disabled=True,
    layout=widgets.Layout(
        width="190px",
        height="50px"
    )
)


# ============================================================
# CHECK ANSWER
# ============================================================

def check_addition(button):

    global level6_score
    global level6_lives

    if len(level6_selected) == 2:

        # ----------------------------------------------
        # Correct
        # ----------------------------------------------

        level6_score += 20

        feedback.value = """
        <div style="
            background:#D9FFD9;
            padding:20px;
            border-radius:15px;
            text-align:center;
            font-size:21px;
        ">

        🎉 <b>Excellent!</b>

        <br><br>

        You selected 2 pieces out of 4.

        <br><br>

        🍕 <b>1/4 + 1/4 = 2/4</b>

        <br><br>

        And:

        <br>

        ⭐ <b>2/4 = 1/2</b>

        <br><br>

        You ate <b>half of the pizza!</b> 🍕

        <br><br>

        ⭐ <b>+20 points</b>

        </div>
        """

        update_score()

        # Disable all piece buttons
        for btn in piece_buttons:
            btn.disabled = True

        check_button.disabled = True

        next_level_button.layout.display = "block"

    else:

        # ----------------------------------------------
        # Wrong / incomplete
        # ----------------------------------------------

        level6_lives -= 1

        update_score()

        if level6_lives <= 0:

            feedback.value = """
            <div style="
                background:#FFCCCC;
                padding:20px;
                border-radius:15px;
                text-align:center;
                font-size:21px;
            ">

            💔 <b>Game Over!</b>

            <br><br>

            You used all your lives.

            </div>
            """

            for btn in piece_buttons:
                btn.disabled = True

            check_button.disabled = True

        else:

            feedback.value = """
            <div style="
                background:#FFF2CC;
                padding:15px;
                border-radius:15px;
                text-align:center;
                font-size:20px;
            ">

            👆 Please select <b>2 pieces</b>.

            </div>
            """


check_button.on_click(check_addition)


# ============================================================
# NEXT LEVEL BUTTON
# ============================================================

next_level_button = widgets.Button(
    description="➡️ Next Level",
    button_style="success",
    layout=widgets.Layout(
        width="180px",
        height="45px"
    )
)

next_level_button.layout.display = "none"


def go_next_level(button):

    feedback.value = """
    <div style="
        background:#D9EAF7;
        padding:20px;
        border-radius:15px;
        text-align:center;
        font-size:22px;
    ">

    🎉 <b>Level 6 Completed!</b>

    <br><br>

    You learned:

    <br><br>

    <b>1/4 + 1/4 = 2/4 = 1/2</b>

    <br><br>

    🍕 Great job, Fraction Chef!

    <br><br>

    🔜 <b>Next: Level 7 – Fraction Subtraction</b>

    </div>
    """

    next_level_button.disabled = True


next_level_button.on_click(go_next_level)


# ============================================================
# ARRANGE PIECE BUTTONS
# ============================================================

piece_row = widgets.HBox(
    piece_buttons,
    layout=widgets.Layout(
        justify_content="center"
    )
)


# ============================================================
# COMPLETE LEVEL UI
# ============================================================

level6_ui = widgets.VBox(
    [

        title,

        score_display,

        question,

        pizza_image,

        status,

        piece_row,

        widgets.HTML(
            "<br>"
        ),

        check_button,

        feedback,

        next_level_button

    ],

    layout=widgets.Layout(
        width="100%",
        align_items="center"
    )
)


# ============================================================
# DISPLAY GAME
# ============================================================

display(level6_ui)
```
